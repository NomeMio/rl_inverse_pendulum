```python
import  cart_model
from tqdm.auto import tqdm
import policies
import random
random.seed(0)
context = cart_model.Cart_model(human=False)
runs=100
```

```python
#Testing random policy
average_reward=0
max_reward=0
min_reward=1000000
for i in range(runs):
    startingState, _ = context.reset()
    episode = cart_model.Episode()
    episode.run(context,None)
    if episode.total_reward>max_reward:
        max_reward=episode.total_reward
    if episode.total_reward<min_reward:
        min_reward=episode.total_reward
    average_reward+=episode.total_reward
average_reward/=runs
print(f"Random policy: average reward: {average_reward}, max reward: {max_reward}, min reward: {min_reward}")
```

```python
average_reward=0
max_reward=0
min_reward=1000000
p=policies.ActorPolicyContinuousSpace.load()
for i in range(runs):
    startingState, _ = context.reset()
    episode = cart_model.Episode()
    episode.run(context,p)
    if episode.total_reward>max_reward:
        max_reward=episode.total_reward
    if episode.total_reward<min_reward:
        min_reward=episode.total_reward
    average_reward+=episode.total_reward
average_reward/=runs
print(f"Random policy: average reward: {average_reward}, max reward: {max_reward}, min reward: {min_reward}")
```

```python
average_reward=0
max_reward=0
min_reward=1000000
p=policies.sarsaWithQuantization.load("saved_policies/sarsa_best_policy")
for i in range(runs):
    startingState, _ = context.reset()
    episode = cart_model.Episode()
    episode.run(context,p)
    if episode.total_reward>max_reward:
        max_reward=episode.total_reward
    if episode.total_reward<min_reward:
        min_reward=episode.total_reward
    average_reward+=episode.total_reward
average_reward/=runs
print(f"Best Sarsa policy: average reward: {average_reward}, max reward: {max_reward}, min reward: {min_reward}")
print(p.getParameters())
```
